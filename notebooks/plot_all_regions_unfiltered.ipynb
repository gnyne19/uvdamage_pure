```python
from pathlib import Path
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

OUT = Path('/cta/users/guneyn23/relative_repair/window_mean_relative_all_regions')
summary = pd.read_csv(OUT / 'all_regions.tsv', sep='\t', dtype={'damage': str})

# Oranı filtresiz olarak yeniden hesapla; önceki NaN filtresini kullanma.
summary['relative_real_div_sim'] = summary['relative_real'] / summary['relative_sim']

REGIONS = ['ATAC', 'H3K9me3', 'H3K27me3']
DAMAGES = ['CPD', '64']
TIMES = ['15m', '30m', '1h', '4h', '8h']
COLORS = {'15m':'#ff595e', '30m':'#ffca3a', '1h':'#8ac926', '4h':'#1982c4', '8h':'#6a4c93'}
x = np.linspace(-10, 10, 400, endpoint=False) + 10/400
```

```python
for region in REGIONS:
    for damage in DAMAGES:
        fig, ax = plt.subplots(figsize=(10, 5))

        for time in TIMES:
            data = summary[
                (summary['region'] == region)
                & (summary['damage'] == damage)
                & (summary['time'] == time)
            ]
            ax.plot(x, data['relative_real_div_sim'], color=COLORS[time], label=time)

        ax.axvline(0, color='gray', linestyle='--')
        ax.set_xlim(-10, 10)
        ax.set_xlabel('Distance from region center (kb)')
        ax.set_ylabel('Relative real / relative sim')
        ax.set_title(f'{region} {damage} — unfiltered')
        ax.legend()
        fig.tight_layout()
        plot_file = OUT / f'{region}_{damage}_unfiltered.png'
        fig.savefig(plot_file, dpi=300)
        plt.show()
        print(plot_file)
```
